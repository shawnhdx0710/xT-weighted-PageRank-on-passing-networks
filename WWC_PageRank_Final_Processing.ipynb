# Final Processing for WPR

The Z-score tells us how many standard deviations a player's performance was from the average performance in that specific match. It normalizes the raw WPR_score while preserving the magnitude of the performance.

A positive Z-score means the player was above average in that match; a negative score means they were below average.

For each match, calculate the mean (μ) and standard deviation (σ) of all WPR_scores. For each player in that match, calculate their Z-Score:

Z-Score = (player_WPR_score - μ) / σ

Aggregate these Z-scores for each player across all their matches by calculating the mean. A consistently high-performing player will have a high positive average Z-score. Rank players based on their average Z-score, from highest to lowest.


```python
import pandas as pd
import numpy as np

try:
    df = pd.read_csv('/Users/shawnhan/Desktop/Pioneer/Pioneer_Final_Project/all_matches_wpr_combined.csv')
    print("Successfully loaded 'all_matches_wpr_combined.csv'.")
except FileNotFoundError:
    print("Error: The file 'all_matches_wpr_combined.csv' was not found.")
    exit()

match_mean = df.groupby('match_id')['WPR_score'].transform('mean')
match_std = df.groupby('match_id')['WPR_score'].transform('std')

df['z_score'] = np.where(match_std > 0, (df['WPR_score'] - match_mean) / match_std, 0)

player_rankings = df.groupby(['player_id', 'player', 'team']).agg(
    average_z_score=('z_score', 'mean'),
    matches_played=('match_id', 'nunique')
).reset_index()

final_ranked_df = player_rankings.sort_values(by='average_z_score', ascending=False)

output_filename = 'player_rankings_z_score.csv'
try:

    final_ranked_df.to_csv(output_filename, index=False)
    
    print(f"\nSuccessfully saved the ranked data to '{output_filename}'.")
    print("This file contains all players, ranked by their average Z-score.")

except Exception as e:
    print(f"\nAn error occurred while saving the file: {e}")

print("\n--- Top 15 Players from the new ranked file ---")
print(final_ranked_df.head(15).to_string())
```
